# CareerLens AI — Phases 6 & 7: Career Recommendation & Skill-Gap Analysis

In this notebook, we evaluate our integrated recommendation and skill-gap engine.

### Core Features Tested:
1. **Input Skill Processing**: Normalization and multi-attribute matching.
2. **Hybrid Scoring**: Balancing Direct Skill Overlap with TF-IDF Cosine Similarity.
3. **Skill-Gap Identification**: Discovering missing tools and prioritizing Hot Technologies to learn.
4. **Target Career Diagnostics**: Assessing readiness for a specific dream role.

## 1. Initialize the CareerRecommender

```python
import sys
import os

# Add src folder to Python module search path
SRC_DIR = os.path.join(os.path.dirname(os.getcwd()), "src")
if SRC_DIR not in sys.path:
    sys.path.append(SRC_DIR)

from recommender import CareerRecommender

recommender = CareerRecommender()
print("CareerRecommender successfully initialized!")
```

## 2. Test Recommendation with Custom Skills
Try changing the list of skills below to test different profiles (e.g. software engineering, data science, project management)!

```python
my_skills = ["Python", "SQL", "Excel", "Statistics", "Data Visualization"]

recommendations = recommender.recommend(my_skills, top_n=5)

for rank, rec in enumerate(recommendations, 1):
    print(f"\n{'='*75}")
    print(f"Rank #{rank}: {rec['title']} (SOC Code: {rec['soc_code']})")
    print(f"Overall Match Score: {rec['match_score']}% [Skill Overlap: {rec['overlap_percentage']}%, TF-IDF: {rec['cosine_similarity']}%]")
    print(f"Skills Matched: {', '.join(rec['matched_skills'])}")
    print(f"Recommended Skills to Learn Next: {', '.join(rec['recommended_skills_to_learn'])}")
```

## 3. Test Target Career Skill-Gap Analysis
What if you have a specific dream role in mind, such as **'Data Scientists'** or **'Software Developers'**?

```python
target_role = "Data Scientists"
gap_report = recommender.analyze_skill_gap(my_skills, target_role)

print(f"Target Occupation: {gap_report['title']}")
print(f"Description: {gap_report['description'][:200]}...")
print(f"Skills You Already Possess: {gap_report['matched_skills']}")
print(f"High-Priority Technologies to Learn: {gap_report['priority_hot_technologies_to_learn']}")
```
