## Step 1: Import Required Libraries
We use **Pandas** for reading tabular data and **OS** to navigate file paths.

```python
import os
import pandas as pd

# Set display settings to see all columns clearly
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 1000)

# Define the path to our data directory
DATA_DIR = os.path.join(os.path.dirname(os.getcwd()), "data")
print(f"Data Directory: {DATA_DIR}")
```

## Step 2: Load and Inspect Dataset 1 — Occupation Master Data (`occupation_data.csv`)
This dataset contains the master list of 1,016 standard occupations, their unique O*NET-SOC codes, and detailed job descriptions.

```python
occ_path = os.path.join(DATA_DIR, "occupation_data.csv")
df_occ = pd.read_csv(occ_path)

print(f"Dimensions: {df_occ.shape[0]:,} rows, {df_occ.shape[1]} columns")
print(f"Missing values:\n{df_occ.isnull().sum()}")
df_occ.head(3)
```

## Step 3: Load and Inspect Dataset 2 — Software Skills (`software_skills.csv`)
This dataset maps each occupation to concrete software tools, programming languages, and technologies.

```python
soft_path = os.path.join(DATA_DIR, "software_skills.csv")
df_soft = pd.read_csv(soft_path)

print(f"Dimensions: {df_soft.shape[0]:,} rows, {df_soft.shape[1]} columns")
print(f"Missing values:\n{df_soft.isnull().sum()}")
print(f"Total unique software skills recorded: {df_soft['Workplace Example'].nunique():,}")
df_soft.head(3)
```

## Step 4: Load and Inspect Dataset 3 — Transferable Skills (`transferable_skills.csv`)
This dataset captures foundational human skills (Critical Thinking, Mathematics, Complex Problem Solving, etc.) scored by Importance (`IM`) and Level (`LV`).

```python
trans_path = os.path.join(DATA_DIR, "transferable_skills.csv")
df_trans = pd.read_csv(trans_path)

print(f"Dimensions: {df_trans.shape[0]:,} rows, {df_trans.shape[1]} columns")
print(f"Missing values:\n{df_trans.isnull().sum()}")
print(f"Unique skill elements: {df_trans['Element Name'].unique()}")
df_trans.head(3)
```

## Step 5: Load and Inspect Dataset 4 — Training & Experience (`training_experience.csv`)
This dataset tracks education requirements (High School, Bachelor's, Master's) and prior years of experience needed for each job.

```python
train_path = os.path.join(DATA_DIR, "training_experience.csv")
df_train = pd.read_csv(train_path)

print(f"Dimensions: {df_train.shape[0]:,} rows, {df_train.shape[1]} columns")
print(f"Missing values:\n{df_train.isnull().sum()}")
df_train.head(3)
```
