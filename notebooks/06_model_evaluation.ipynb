# CareerLens AI — Phase 9: Model Evaluation & Testing

In this notebook, we systematically evaluate our Recommendation System.

### Why Standard Classification Accuracy Doesn't Apply:
In classification (e.g., Cat vs. Dog), there is a single correct label. But in recommendation systems, an engineer with Python, SQL, and Statistics could thrive as a **Data Scientist**, a **Biostatistician**, or a **Quantitative Analyst**.

Therefore, we evaluate using **Information Retrieval Ranking Metrics**:
1. **Precision@K**: The fraction of the top $K$ recommendations that belong to the relevant career domain.
2. **Mean Reciprocal Rank (MRR)**: Measures how close to Rank #1 the first relevant occupation appears.

## 1. Setup & Imports

```python
import sys
import os
import numpy as np
import pandas as pd

SRC_DIR = os.path.join(os.path.dirname(os.getcwd()), "src")
if SRC_DIR not in sys.path:
    sys.path.append(SRC_DIR)

from recommender import CareerRecommender

recommender = CareerRecommender()
print("CareerRecommender ready for evaluation.")
```

## 2. Multi-Domain Benchmark Evaluation

```python
benchmark = [
    {
        "persona": "Data Science & Analytics",
        "skills": ["Python", "SQL", "Machine Learning", "Statistics", "Tableau", "Pandas"],
        "relevant_domains": ["Data Scientist", "Statistic", "Research Scientist", "Quantitative Analyst", "Analytics"]
    },
    {
        "persona": "Software Engineering",
        "skills": ["Java", "C++", "Python", "Git", "Linux", "Software Architecture", "Docker"],
        "relevant_domains": ["Software Developer", "Programmer", "Blockchain Engineer", "Systems Engineer", "Architect"]
    },
    {
        "persona": "Web & UI Development",
        "skills": ["JavaScript", "HTML", "CSS", "React", "Node.js", "Git", "Web browser software"],
        "relevant_domains": ["Web Developer", "Interface Designer", "Software Developer", "Programmer"]
    }
]

for b in benchmark:
    recs = recommender.recommend(b["skills"], top_n=5)
    titles = [r["title"] for r in recs]
    hits = sum(1 for t in titles if any(d.lower() in t.lower() for d in b["relevant_domains"]))
    print(f"{b['persona']} -> Precision@5: {hits/5:.2f} ({hits}/5 relevant)")
    print(f"   Recommendations: {titles}\n")
```

## 3. Edge-Case Robustness Testing

```python
# Empty input check
print("Empty input output:", recommender.recommend([]))

# Case-insensitivity check
rec_lower = recommender.recommend(["python", "sql"], top_n=1)[0]["title"]
rec_upper = recommender.recommend(["PYTHON", "SQL"], top_n=1)[0]["title"]
print(f"Case matching identical: {rec_lower == rec_upper} ('{rec_lower}')")
```
